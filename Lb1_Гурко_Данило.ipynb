### Завдання 28

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mean = np.mean (A, axis=0)
std = np.std(A, axis=0)

normalized = (A - mean) / std

print(normalized)
````

Дано результати продажів трьох товарів за чотири місяці:

```text
sales = [[120, 150, 170, 160],
         [80,  90,  100, 110],
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
# Код студента
sales = [
    [120, 150, 170, 160],
    [80, 90, 100, 110],
    [200, 180, 210, 230]
]

#Загальний обсяг продажів кожного товару
product_totals = [sum(product) for product in sales]
print("1. Продажі кожного товару:", product_totals)

#Загальний обсяг продажів у кожному місяці
month_totals = [sum(sales[i][j] for i in range(3)) for j in range(4)]
print("2. Продажі кожного місяця:", month_totals)

#Товар із найбільшими сумарними продажами
max_product = product_totals.index(max(product_totals)) + 1
print("3. Товар з найбільшими продажами:", max_product)

#Місяць із найбільшим обсягом продажів
max_month = month_totals.index(max(month_totals)) + 1
print("4. Місяць з найбільшими продажами:", max_month)

#Середній обсяг продажів кожного товару
averages = [sum(product) / len(product) for product in sales]
print("5. Середні продажі кожного товару:", averages)
````

### Завдання 37
Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента
import random

# Створюємо матрицю 5x5 з випадкових чисел від 1 до 20
matrix = [[random.randint(1, 20) for _ in range(5)] for _ in range(5)]

print("Початкова матриця:")
for row in matrix:
    print(row)

# Сума головної діагоналі
main_diagonal_sum = sum(matrix[i][i] for i in range(5))
print("\n1. Сума головної діагоналі:", main_diagonal_sum)

# Сума побічної діагоналі
secondary_diagonal_sum = sum(matrix[i][4 - i] for i in range(5))
print("2. Сума побічної діагоналі:", secondary_diagonal_sum)

# Максимальний елемент кожного рядка
row_max = [max(row) for row in matrix]
print("3. Максимальні елементи рядків:", row_max)

# Мінімальний елемент кожного стовпця
column_min = [min(matrix[i][j] for i in range(5)) for j in range(5)]
print("4. Мінімальні елементи стовпців:", column_min)

# Заміна елементів, менших за середнє значення, на 0
average = sum(sum(row) for row in matrix) / 25

for i in range(5):
    for j in range(5):
        if matrix[i][j] < average:
            matrix[i][j] = 0

print("\n5. Середнє значення матриці:", average)
print("Матриця після заміни:")
for row in matrix:
    print(row)
````

### Завдання 38
Дано матрицю ознак трьох об’єктів:

```text
X = [[2, 4, 6],
     [1, 3, 5],
     [5, 7, 9]]
```

Обчисліть **евклідову відстань** між:
- першим і другим рядком;
- першим і третім рядком;
- другим і третім рядком.

Використайте операції NumPy та `np.linalg.norm()`.

````python
# Код студента
import numpy as np

X = np.array([
    [2, 4, 6],
    [1, 3, 5],
    [5, 7, 9]
])

d12 = np.linalg.norm(X[0] - X[1])
d13 = np.linalg.norm(X[0] - X[2])
d23 = np.linalg.norm(X[1] - X[2])

print("Відстань між 1 і 2:", d12)
print("Відстань між 1 і 3:", d13)
print("Відстань між 2 і 3:", d23)
````

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
# Код студента
import numpy as np

X = np.array([
    [1, 2, 3, 4],
    [2, 4, 6, 8],
    [1, 3, 5, 7],
    [3, 5, 7, 9]
])

# Норма кожного рядка
norms = np.linalg.norm(X, axis=1)

# Нормалізація кожного рядка до одиничної довжини
X_normalized = X / norms[:, np.newaxis]

print("Початкова матриця:")
print(X)

print("\nНорми рядків:")
print(norms)

print("\nНормалізована матриця:")
print(X_normalized)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
import numpy as np

# Матриця коефіцієнтів
A = np.array([
    [1,  2,  1],
    [2,  1, -1],
    [3, -1,  2]
])

# Вектор правих частин
b = np.array([9, 8, 3])

# Розв'язання системи
x = np.linalg.solve(A, b)

print("Розв'язок системи:")
print(x)

# Перевірка матричним множенням A @ x
result = A @ x

print("\nA @ x:")
print(result)

# Порівняння з b
print("\nВектор b:")
print(b)

print("\nЧи дорівнює A @ x вектору b?")
print(np.allclose(result, b))
````
